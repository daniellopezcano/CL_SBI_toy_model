# 01 · Generate the datasets

Toy model with cosmological parameters **A, B** and the baryonic parameter **C**. Two baryonic models are used
for training, which differ only in the range of C: **v1** (C ∈ [−2, −1.7]) and **v2** (C ∈ [−0.3, 0]).
Both use the same cosmologies, so the v1 and v2 spectra of one cosmology are positive pairs for CL.

Writes to `data/<run>/`:
* `pk/train_v1.npz`, `pk/train_v2.npz`, `pk/val_v1.npz`, `pk/val_v2.npz`: LHS cosmologies, 2 C values each;
* `pk/test_grid.npz`: LHS cosmologies × a regular grid of C over its full prior [−2, 2];
* `norm.npz`: per-k mean/std (computed on a set with C over the full prior) used to normalise all spectra.

Every `.npz` has `cosmos (n_cosmo, 2)`, `aug (n_cosmo, n_aug, 1)` and `xx (n_noise, n_cosmo, n_aug, n_k)`.
